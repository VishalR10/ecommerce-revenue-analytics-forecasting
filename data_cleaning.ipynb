```python
import pandas as pd

# Read CSV using latin1 encoding
df = pd.read_csv(
    "data/superstore.csv",
    encoding="latin1"
)

# Show first rows
print(df.head())

# Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

# Remove hidden special characters
for col in df.select_dtypes(include="object").columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("\xa0", " ", regex=False)
        .str.encode("ascii", "ignore")
        .str.decode("ascii")
    )

# Save cleaned CSV
df.to_csv(
    "data/superstore_mysql_clean.csv",
    index=False,
    encoding="ascii"
)

print("Clean CSV created successfully!")
print(df.columns)
```
