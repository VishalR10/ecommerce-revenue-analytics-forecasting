# E-Commerce Revenue Analytics and Forecasting

## Python Exploratory Data Analysis (EDA)

This notebook analyzes retail sales transactions to identify business insights related to revenue, profitability, regional performance, product performance, discount impact, and seasonal sales trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loading the dataset

file_path = '/Users/vishalrathi/Desktop/ecommerce-revenue-analytics-forecasting/data/superstore_mysql_clean.csv'
df = pd.read_csv(file_path)
df.head(5)
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'])
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[['Sales','Profit','Quantity','Discount']].describe()
```

# Business KPI Analysis

```python
# Key Business KPI

total_revenue = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order_ID'].nunique()

average_order_value = total_revenue / total_orders

profit_margin = (total_profit/total_revenue) * 100
```

```python
print("Total Revenue:", round(total_revenue, 2))

print("Total Profit:", round(total_profit, 2))

print("Total Orders:", total_orders)

print("Average Order Value:", round(average_order_value, 2))

print("Profit Margin %:", round(profit_margin, 2))
```

# Regional Performance Analysis

```python
region_summary = (
    df.groupby("Region")
    .agg(
        total_revenue=("Sales", "sum"),
        total_profit=("Profit", "sum"),
        total_orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values(by="total_revenue", ascending=False)
)

region_summary
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    region_summary["Region"],
    region_summary["total_revenue"]
)

plt.title("Revenue by Region")

plt.xlabel("Region")

plt.ylabel("Revenue")

plt.tight_layout()

plt.savefig("../images/revenue_by_region.png")

plt.show()
```

### Business Insight

The West region generated the highest revenue and profit, making it the strongest regional market. Central generated significant revenue but comparatively lower profit, suggesting possible margin or discount-related inefficiencies.

# Category Performance Analysis

```python
category_summary = (
    df.groupby("Category")
    .agg(
        total_revenue=("Sales", "sum"),
        total_profit=("Profit", "sum"),
        total_quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values(by="total_revenue", ascending=False)
)

category_summary
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    category_summary["Category"],
    category_summary["total_revenue"]
)

plt.title("Revenue by Category")

plt.xlabel("Category")

plt.ylabel("Revenue")

plt.tight_layout()

plt.savefig("../images/revenue_by_category.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    category_summary["Category"],
    category_summary["total_profit"]
)

plt.title("Profit by Category")

plt.xlabel("Category")

plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../images/profit_by_category.png")

plt.show()
```

### Business Insight

Technology generated the highest revenue and profit, making it the strongest business category. Office Supplies sold the highest quantity of products, indicating high transaction volume but lower-value items. Furniture generated strong revenue but relatively weak profit, suggesting possible operational or discount-related inefficiencies.

# Monthly Revenue Trend Analysis

```python
df["Order_Month"] = df["Order_Date"].dt.to_period("M").astype(str)

df[["Order_Date", "Order_Month"]].head()
```

```python
monthly_summary = (
    df.groupby("Order_Month")
    .agg(
        monthly_revenue=("Sales", "sum"),
        monthly_profit=("Profit", "sum")
    )
    .reset_index()
)

monthly_summary.head()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_summary["Order_Month"],
    monthly_summary["monthly_revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.xticks(rotation=90)

plt.tight_layout()

plt.savefig("../images/monthly_revenue_trend.png")

plt.show()
```

### Business Insight

Monthly revenue trends show clear seasonality patterns across the business. Revenue spikes are visible toward year-end, especially during late 2016, suggesting strong holiday or year-end purchasing behavior. Seasonal trends like these are highly important for forecasting, inventory planning, and marketing strategy.

# Product Performance Analysis

```python
top_products_revenue = (
    df.groupby("Product_Name")
    .agg(
        total_revenue=("Sales", "sum")
    )
    .reset_index()
    .sort_values(by="total_revenue", ascending=False)
    .head(10)
)

top_products_revenue
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    top_products_revenue["Product_Name"],
    top_products_revenue["total_revenue"]
)

plt.title("Top 10 Products by Revenue")

plt.xlabel("Revenue")

plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("../images/top_10_revenue_products.png")

plt.show()
```

```python
top_products_profit = (
    df.groupby("Product_Name")
    .agg(
        total_profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(by="total_profit", ascending=False)
    .head(10)
)

top_products_profit
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    top_products_profit["Product_Name"],
    top_products_profit["total_profit"]
)

plt.title("Top 10 Products by Profit")

plt.xlabel("Profit")

plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("../images/top_10_product_by_profit.png")

plt.show()
```

### Business Insight

Several technology and office equipment products dominate both revenue and profitability rankings. High-ticket products such as copiers and printers generate substantial business value despite lower transaction frequency. This suggests that premium products contribute significantly to overall profitability and should be strategically prioritized.

# Loss-Making Product Analysis

```python
loss_products = (
    df.groupby("Product_Name")
    .agg(
        total_profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(by="total_profit", ascending=True)
    .head(10)
)

loss_products
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    loss_products["Product_Name"],
    loss_products["total_profit"]
)

plt.title("Top 10 Loss-Making Products")

plt.xlabel("Profit")

plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("../images/top_10_loss-making_products.png")

plt.show()
```

### Business Insight

Several products generate significant negative profit despite contributing to overall sales. Large furniture items and specialized office equipment appear particularly unprofitable, potentially due to excessive discounting, shipping costs, or operational inefficiencies. These products should be reviewed for pricing optimization, supplier renegotiation, or inventory strategy adjustments.

# Discount vs Profit Analysis

```python
discount_profit = df[["Discount", "Profit", "Sales"]]

discount_profit.head()
```

```python
discount_profit.corr()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Discount"],
    df["Profit"],
    alpha=0.4
)

plt.title("Discount vs Profit")

plt.xlabel("Discount")

plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../images/discount_vs_profit_correlation.png")

plt.show()
```

### Business Insight

The relationship between discount and profit appears negatively correlated, indicating that higher discounts often reduce profitability. While discounting may help drive sales volume, excessive discounts can significantly impact margins and create loss-making transactions. Businesses should carefully optimize discount strategies to balance growth and profitability.
